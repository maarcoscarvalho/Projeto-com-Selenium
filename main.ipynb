```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

servico = Service(ChromeDriverManager().install())
navegador = webdriver.Chrome(service=servico)
arquivo = "file:///Users/marcosdelbem/Desktop/Selenium/Pagina%20Hashtag.html"
navegador.get(arquivo)
```

```python
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait

WebDriverWait(navegador, 10).until(
    lambda driver: driver.execute_script("return document.readyState") == "complete"
)

elementos_logo = navegador.find_elements(
    By.CSS_SELECTOR,
    ".custom-logo, .custom-logo-link"
)

campo_nome = elementos_logo[0] if elementos_logo else None

if campo_nome is not None:
    campo_nome.click()
```
